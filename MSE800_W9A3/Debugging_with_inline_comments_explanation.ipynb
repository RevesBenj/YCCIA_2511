```python
# EXAMPLE 2 - LIST INDEX OUT OF RANGE FIX

import random

# ISSUE:
# List index start at 0.
# randint(1,6) can return 6.
# dice_images[6] cause index error (out of range).

dice_images = [1, 2, 3, 4, 5, 6]
dice_num = random.randint(1, 6)

# FIX:
# minus 1 to match correct index

print(dice_images[dice_num - 1])

#  OUTPUT:
# random number from 1 to 6
```
